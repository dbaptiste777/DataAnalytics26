 ## Generating Random Choices from an array

```python
students = np.array ([
    "naomi", "victoria", "luis", "jennifer",
    "smaya", "joseph", "carter", "june",
    "elias", "frank", "hector", "jose"
])

print(students)
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students,(3,4)) #This wouldnt be a good way to create breakout rooms because some of these names are duplicated and its not possible for one person to be in two breakout rooms.
print(breakout_groups)
```

```python
call_on_weighted = np.random.choice(
    students,
    5,
    p=[0.30,0.10,0.10,0.05,0.05,0.1,0.00,0.05,0.05,0.05, .1,.05]
)
print(call_on_weighted)
```
