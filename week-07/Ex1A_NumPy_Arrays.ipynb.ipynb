## Looping Through Arrays

```python
import numpy as np
```

```python
arrA = np.array([
    [11, 12, 13],
    [14,15,16]
                ])
print(arrA)
```

```python
for x in arrA:  #Loop through first dimension
    print(x * 2)
```

```python
for x in arrA:    #loop through second dimension
    for y in x:
        print (y * 2)
```

## The main difference I notice for this example is that in the first loop, the numbers are grouped together in brackets, seperated by commas. The second loop just looks like a list of numbers.

```python
loop_result = x * 2 #loop through second dimension with x * 2

for x in arrA:
    for y in x:
        loop_result = x * 2
        print (loop_result)  #The main difference that I notice with this script is that the x * 2 doubles the whole row where as y * 2 doubles one number itself.
```

## Joining and Stacking Arrays

```python
import numpy as np
```

```python
arrA = np.array([[11,12,13],
                [14,15,16]])
```

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0) #axis = 0
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1) #axis = 1
print(joined_array_2)
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA, axis==0))   #Error happened because axis = 1 needs both arrays 
print(another_joined_array)
```

```python
print(arrA.shape)
print(joined_array_1.shape)
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA), axis=0)  #Error happened because axis = 0. When trying to concatenate an array on a default axis the columns must match. 
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis = 1) #This was able to run when the axis=1 because both arrays have the same amount of rows. 
print(another_2)
```

```python
code_red = np.vstack((arrA,arrA)) #Stacks arrays on top eachother
print(code_red)
```

```python
code_black = np.hstack((arrA,arrA)) #Stacks arrays side by side
print(code_black)
```

```python
code_purple = np.dstack((arrA,arrA)) #Stacks arrays in 3 dimensions.
print(code_purple)
```

## Sorting and Filtering Array Values.

```python
import numpy as np
```

```python
arrB = np.array([ #Create variable 
    [7,6],
    [13,5],
    [7,3]
])
print(arrB)
    
```

```python
arrB_sorted = np.sort(arrB) #Sorted
print(arrB_sorted)
```
